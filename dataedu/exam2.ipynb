````python
import requests
from bs4 import BeautifulSoup


url = "https://novel.naver.com"
print("***네이버 웹소설 통합 랭킹***\n")

response = requests.get(url) # 웹소설 랭킹 가져오기
if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")
    ranking_items = soup.select("span.title")
 
    for rank, item in enumerate(ranking_items, 1): #웹소설 통합 랭킹 출력 코드      
        if item:
            title = item.get_text(strip=True)
            print(f"[{rank}위] {title}")    
else:
    
    print(f"페이지를 불러오지 못했습니다. (에러 코드: {response.status_code})")

print("\n수집이 완료되었습니다.")
````
