## Series — 1차원 데이터
`Series`: **인덱스가 붙은 1차원 배열** DataFrame의 각 열(column)이 Series 임

```python
# 리스트로 생성
import pandas as pd
scores = pd.Series([85, 92, 78, 96, 88])
print(scores)
print(f"\n평균: {scores.mean():.1f}  최대: {scores.max()}  최소: {scores.min()}")
```

```python
print(scores.index)
```

```python
print(scores[0])     # 라벨기반 접근
print(scores.loc[0]) # 라벨기반 접근
print(scores.iloc[0]) # 위치기반 접근
```

```python
# 리스트로 생성
scores = pd.Series([85, 92, 78, 96, 88], index=[10, 20, 30, 40, 50])
print(scores)
print(f"\n평균: {scores.mean():.1f}  최대: {scores.max()}  최소: {scores.min()}")
```

```python
print(scores.index)
```

```python
print(scores[10])     # 라벨기반 접근
print(scores.loc[10]) # 라벨기반 접근
print(scores.iloc[0]) # 위치기반 접근
```

```python
# 딕셔너리로 생성 — 키 : 인덱스, 값 : 데이터 값
fruit_prices = pd.Series({'사과': 1500, '바나나': 800, '포도': 3000, '딸기': 4500})
print(fruit_prices)
```

```python
print("인덱스:", fruit_prices.index.tolist())
print("값:    ", fruit_prices.values)
```

```python
print(fruit_prices.index)
```

```python
# 인덱스로 접근
print("사과 가격:", fruit_prices['사과'])
print("사과 가격:", fruit_prices.loc['사과'])
print("사과 가격:", fruit_prices.iloc[0])
```

```python
print(fruit_prices >= 1500)
```

```python
print(fruit_prices + 1500)
```

```python
print("1500원 이상:", fruit_prices[fruit_prices >= 1500], sep="\n")
```

```python
exam_data = {'name' : [ 'dooly', 'ddoch', 'dounar'],
             'math' : [ 90, 80, 70],
             'eng' : [ 98, 89, 95],
             'music' : [ 85, 95, 100],
             'exer' : [ 100, 90, 90]}
df = pd.DataFrame(exam_data)
print(df)
df.set_index('name', inplace=True)
print(df)
```

```python
df.plot()
```

```python
# 데이터프레임 df를 전치하기 (메소드 활용)
df = df.transpose()
print(df)
```

```python
df.plot()
```

```python
# 데이터프레임 df를 다시 전치하기 (클래스 속성 활용)
df = df.T
print(df)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'name':['둘리', '또치', '도우너', '희동이'],
    '국어':[90, 80, 70, 70],
    '영어':[99, 98, 97, 46],
    '수학':[90, 70, 70, 60],
}
df = pd.DataFrame(data)
display(df)
df.set_index('name', inplace=True)
display(df)
```

```python
import matplotlib.pyplot as plt
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False 
```

```python
df.plot()
plt.show()
```

```python
df.T.plot()
plt.show()
```

```python
df.plot(kind='bar')
plt.show()
```

```python
df.plot(kind='bar', stacked=True)
plt.show()
```

```python
df.plot(kind='barh')
plt.show()
```

```python
df.plot(kind='box')
plt.show()
```

```python
df.plot(kind='pie',y='국어', legend=False, ylabel="", title="학생들의 국어 점수 파이플롯")
plt.show()
```

```python
df["총점"] = df.국어+df.영어+df.수학
df
```

```python
#총점
#colormap=plt.cm.gist_rainbow 정해둔 컬러를 설정
df.plot(kind='pie',y='총점', legend=False, ylabel="", title="학생들의 총점 파이플롯", colormap=plt.cm.gist_rainbow)
plt.show()
```
