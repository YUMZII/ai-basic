## 조건 필터링
**boolean indexing**을 활용해 원하는 행만 골라냅니다.

```python
# 단일 조건
print("개발부서 직원:")
df[df['부서'] == '개발']
```

```python
df['부서'] == '개발'
```

```python
df[[True, False, True, False, True, False, False, False]]
```

```python
# 복합 조건: & (AND), | (OR), ~ (NOT)
# 연봉 5000 이상 AND 재직중
high_pay_active = df[(df['연봉'] >= 5000) & (df['재직중'] == True)]
print(f"연봉 5000만원 이상 재직자: {len(high_pay_active)}명")
```

```python
high_pay_active
```

```python
# isin — 여러 값 중 하나
df[df['부서'].isin(['개발', '디자인'])]
```

```python
# between — 범위 조건
df[df['나이'].between(25, 34)]
```

### pandas 의 DataFrame.query()
문자열 기반 조건식으로 데이터를 필터링하는 메서드
(SQL의 WHERE 절처럼 쓸 수 있어서 가독성이 좋은 게 특징)

```python
# query 문법 (SQL처럼 직관적)
df.query("연봉 > 4500 and 부서 != '인사'")
```

```python
df.query("나이 > 30")  # df[df['나이'] > 30]
```

```python
min_age = 30
df.query("나이 > @min_age")
```

```python
import pandas as pd

# 파일경로를 찾고, 변수 file_path에 저장
file_path = './data/read_csv_sample.csv'

# read_csv() 함수로 데이터프레임 변환. 변수 df1에 저장
# header 속성을 지정하지 않으면 header=infer 가 사용됨
# "pandas가 파일의 내용을 보고 알아서 추론(인증)하겠다"라는 의미
df1 = pd.read_csv(file_path) 
#제목에 해당하는 내용이 마지막 행에 있다면 우리가 지정해줘야 됨
#기본적으로 첫번째 행이 제목이 되는 경우임
display(df1) #마지막에 ,,9면 결측치? 
```

```python
# read_csv() 함수로 데이터프레임 변환. 변수 df2에 저장. header=None 옵션
# header=None은 "이 파일(또는 표)에는 컬럼 이름(헤더)이 없으니, 첫 번째 줄부터 전부 실제 데이터로 읽어라"라는 의미
df2 = pd.read_csv(file_path, header=None)
display(df2)
```

```python
# read_csv() 함수로 데이터프레임 변환. 변수 df3에 저장. index_col=None 옵션
# index_col=None은 "파일에 있는 어떤 열(Column)도 인덱스(행 이름)로 쓰지 않겠다"는 의미
df3 = pd.read_csv(file_path, index_col=None)
display(df3)
```

```python
# read_csv() 함수로 데이터프레임 변환. 변수 df4에 저장. index_col='c0' 옵션
df4 = pd.read_csv(file_path, index_col='c0')
display(df4)
```

```python
import pandas as pd

# HTML 파일 경로 or 웹 페이지 주소를 url 변수에 저장
url ='./data/sample.html'

# HTML 웹페이지의 표(table)를 가져와서 데이터프레임으로 변환 
tables = pd.read_html(url)

# 표(table)의 개수 확인
print(len(tables), "개")
```

```python
# tables 리스트의 원소를 iteration하면서 각각 화면 출력
for i in range(len(tables)):
    print("tables[%s]" % i)
    print(tables[i])
    print('\n')
```

```python
# 파이썬 패키지 정보가 들어 있는 두 번째 데이터프레임을 선택하여 df 변수에 저장
df = tables[1] 

# 'name' 열을 인덱스로 지정
df.set_index(['name'], inplace=True)
print(df)
```

```python
import pandas as pd

# 판다스 DataFrame() 함수로 데이터프레임 변환. 변수 df에 저장 
data = {'name' : [ 'Jerry', 'Riah', 'Paul'],
        'algol' : [ "A", "A+", "B"],
        'basic' : [ "C", "B", "B+"],
        'c++' : [ "B+", "C", "C+"],
        }

df = pd.DataFrame(data)
df.set_index('name', inplace=True)   #name 열을 인덱스로 지정
print(df)

# to_csv() 메소드를 사용하여 CSV 파일로 내보내기. 파열명은 df_sample.csv로 저장
df.to_csv("./output/df_sample.csv")
```

```python
import pandas as pd
df = pd.read_excel('./data/남북한발전전력량.xlsx')  # 데이터프레임 변환 
display(df)
```

```python
df_ns = df.iloc[[0, 5], 2:]            # 남한, 북한 발전량 합계 데이터만 추출
df_ns.index = ['South','North']        # 행 인덱스 변경
df_ns.columns = df_ns.columns.map(int) # 열 이름의 자료형을 정수형으로 변경
display(df_ns)
```

```python
df_ns.plot(kind='bar')
```

```python
# 행, 열 전치하여 막대 그래프 그리기
tdf_ns = df_ns.T
print(tdf_ns.head())
```

```python
tdf_ns.plot(kind='bar')
```

```python
# 행, 열 전치하여 히스토그램 그리기
tdf_ns = df_ns.T.astype('int64')
```

```python
tdf_ns.info()
```
