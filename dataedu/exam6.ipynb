### groupby — 집계
- SQL의 `GROUP BY`와 동일한 개념으로 pandas에서 **가장 강력한 기능** 중 하나이다.
- 데이터를 특정 기준으로 묶어서(그룹화해서) 집계/연산을 수행하는 핵심 메서드이다.

```python
# 부서별 평균 연봉
print("부서별 평균 연봉:")
print(df.groupby('부서')['연봉'].mean().sort_values(ascending=False))
```

```python
# 여러 집계 함수 동시 적용 agg
df.groupby('부서')['연봉'].agg(['mean', 'min', 'max', 'count']).round(0)
```

```python
# 여러 열에 집계
df.groupby('부서').agg(
    평균연봉=('연봉', 'mean'),
    인원수=('이름', 'count'),
    평균나이=('나이', 'mean'),
).round(1)
```

```python
df2 = df.copy()
```

```python
# transform — 그룹별 연산 결과가 원본 행마다 확장(Broadcasting)됨
df2 = df.copy()

df2['부서평균연봉'] = df.groupby('부서')['연봉'].transform('mean')
#그룹 연산의 결과지만, 그 행마다 부서 평균 연봉이 추가
df2
```

```python
df2['연봉편차'] = df['연봉'] - df2['부서평균연봉']
df2[['이름', '부서', '연봉', '부서평균연봉', '연봉편차']].sort_values('부서')
```

### Merge / Join
두 DataFrame을 공통 키(key) 기준으로 합칩니다. SQL JOIN과 동일합니다.

```python
# 직원 기본 정보
df_employees = pd.DataFrame({
    '사번': [101, 102, 103, 104, 105],
    '이름': ['둘리', '고길동', '또치', '도우너', '마이콜'],
    '부서코드': ['D01', 'D02', 'D01', 'D03', 'D01'],
})

# 부서 정보
df_dept = pd.DataFrame({
    '부서코드': ['D01', 'D02', 'D03', 'D04'],
    '부서명': ['개발', '마케팅', '인사', '디자인'],
    '층수': [3, 5, 7, 4],
})

print("직원 테이블:"); display(df_employees)
print("\n부서 테이블:"); display(df_dept)
```

```python
# INNER JOIN (default)
merged_inner = pd.merge(df_employees, df_dept, on='부서코드')
print("INNER JOIN:")
merged_inner
```

```python
# RIGHT JOIN — 부서 테이블 기준, 매칭 안 되는 부서도 포함
merged_right = pd.merge(df_employees, df_dept, on='부서코드', how='right')
print("RIGHT JOIN:"); 
merged_right
#결측치 NaN
```

```python
# concat — 데이터 위아래로 이어 붙이기
df_a = pd.DataFrame({'이름': ['A', 'B'], '점수': [80, 90]})
df_b = pd.DataFrame({'이름': ['C', 'D'], '점수': [70, 85]})
combined = pd.concat([df_a, df_b], ignore_index=True)
print(combined)
```

```python
# concat — 옆으로 이어 붙이기
df_a = pd.DataFrame({'이름': ['A', 'B'], '점수': [80, 90]})
df_b = pd.DataFrame({'이름': ['C', 'D'], '점수': [70, 85]})
display(df_a)
display(df_b)
combined = pd.concat([df_a, df_b], ignore_index=True, axis=1)#axis=1 가로축, 행
display(combined)
```

### Apply / Map / Lambda
복잡한 변환 로직을 열 단위, 행 단위, 또는 요소 단위로 적용

```python
# 딕셔너리로 생성
import pandas as pd
data = {
    '이름': ['둘리', '고길동', '또치', '도우너', '마이콜', '희동이', '엘사', '안나'],
    '나이': [28, 34, 22, 41, 29, 36, 25, 31],
    '부서': ['개발', '마케팅', '개발', '인사', '개발', '마케팅', '디자인', '인사'],
    '연봉': [4800, 5200, 3600, 6100, 4500, 5800, 4200, 5500],
    '입사년도': [2019, 2015, 2023, 2010, 2020, 2013, 2022, 2017],
    '재직중': [True, True, True, False, True, True, False, True],
}

df = pd.DataFrame(data)
display(df)
```

```python
df3 = df[['이름', '나이', '연봉']].copy()
df3
```

```python
# map — Series 원소 하나하나에 적용
df3['나이대'] = df3['나이'].map(lambda x: f"{(x//10)*10}대")
df3
```

```python
df[['이름', '나이', '연봉']].copy()['나이'].map(lambda x: f"{(x//10)*10}대")
```

```python
df[['이름', '나이', '연봉']].copy()['나이'].apply(lambda x: f"{(x//10)*10}대")
```

```python
# apply — Series 또는 DataFrame에 함수 적용
def salary_grade(sal):
    if sal >= 6000: return 'S등급'
    elif sal >= 5000: return 'A등급'
    elif sal >= 4500: return 'B등급'
    else: return 'C등급'

df3['등급'] = df3['연봉'].map(salary_grade) # apply 도 가능
df3
```

```python
# apply — 행(axis=1) 단위로 적용
df3['요약'] = df3.apply(lambda row: f"{row['이름']}({row['나이대']}) - {row['등급']}", axis=1) # map은 불가
print(df3['요약'].tolist())
```

```python
# applymap (pandas 2.x: map) — DataFrame 전체 원소에 적용
numeric_df = df[['나이', '연봉']].copy()
styled = numeric_df.map(lambda x: f"{x:,}")  # 천단위 콤마
print(styled)
```

### 온라인 쇼핑몰의 **주문 데이터**를 분석

| 분석 과제 |
|-----------|
| ① 카테고리별 총 매출과 평균 주문금액 |
| ② 월별 주문 건수 추이 |
| ③ 환불율이 가장 높은 카테고리 |
| ④ VIP 고객 (총 구매 50만원 이상) 목록 |


```python
import pandas as pd
import numpy as np
# 데이터 생성
np.random.seed(7)
n = 200

categories = ['전자기기', '의류', '식품', '도서', '스포츠']
customers = [f'CUST{i:03d}' for i in range(1, 31)]

orders = pd.DataFrame({
    '주문번호': [f'ORD{i:04d}' for i in range(1, n+1)],
    '고객ID': np.random.choice(customers, n),
    '카테고리': np.random.choice(categories, n, p=[0.3, 0.25, 0.2, 0.15, 0.1]),
    '주문금액': np.random.randint(5000, 150000, n),
    '주문일': pd.date_range('2026-01-01', periods=n, freq='2D'),
    '환불여부': np.random.choice([True, False], n, p=[0.12, 0.88]),
})

print(f"주문 데이터: {orders.shape[0]}건")
orders#.head()
```

```python
# (1) 카테고리별 총 매출 & 평균 주문금액
cat_summary = orders.groupby('카테고리').agg(
    총매출=('주문금액', 'sum'),
    주문건수=('주문번호', 'count'),
    평균주문금액=('주문금액', 'mean'),
).sort_values('총매출', ascending=False).round(0)

print("(1) 카테고리별 매출 요약")
display(cat_summary)
```

```python
# (2) 월별 주문 건수 추이
orders['주문월'] = orders['주문일'].dt.to_period('M')
monthly_orders = orders.groupby('주문월')['주문번호'].count()

monthly_orders.plot(kind='bar', figsize=(10, 4), color='steelblue',
                    title='월별 주문 건수', ylabel='주문 건수', edgecolor='white')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
print(f"\n② 평균 월 주문수: {monthly_orders.mean():.1f}건")
```

```python
# (3) 환불율이 가장 높은 카테고리
refund_rate = orders.groupby('카테고리')['환불여부'].mean().sort_values(ascending=False)
refund_rate_pct = (refund_rate * 100).round(1)
print("(3) 카테고리별 환불율(%)")
print(refund_rate_pct)
print(f"\n→ 가장 높은 환불율: {refund_rate_pct.idxmax()} ({refund_rate_pct.max()}%)")
```

```python
# (4) VIP 고객 (총 구매 50만원 이상)
customer_total = orders.groupby('고객ID')['주문금액'].sum().sort_values(ascending=False)
vip = customer_total[customer_total >= 500000]
print(f"(4) VIP 고객 ({len(vip)}명):")
print(vip.reset_index().rename(columns={'주문금액': '총구매액'}))
```
