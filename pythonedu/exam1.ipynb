<span style="color:red; font-weight:bold">데이터 타입(Data type)</span>이란 해당 데이터가 메모리에 어떻게 저장되고, 프로그램에서 어떻게 처리되어야 하는지를 명시적으로 알려주는 것이다.

<span style="color:red; font-weight:bold">변수(variable)</span>란 데이터(data)를 저장하기 위해 프로그램에 의해 이름을 할당받은 메모리 공간을 의미한다.

---

## Primitive(기본, 원시) 타입

```python
#리터럴(데이터값)들의 타입 출력
print(type(100))  
print(type(100.0))
print(type('100'))
print(type("100"))
print(type(True))
print(type(False))
```

```python
print(10)
print(0x10) # 16 * 1 
print(0o10) # 8 * 1
print(0b10) # 2 * 1
```

```python
v1 = 10
v2 = float(10)
v3 = str(10)
v4 = bool(10) # 0 거짓(False), 0 외의 값 참(True)
print(v1)
print(v2)
print(v3)
print(v4)
print()
print(type(v1))
print(type(v2))
print(type(v3))
print(type(v4))
```

```python
print(25+10)        # 덧셈
print('25'+'10')    # 문자열 결합
print(25-10)
print(25*10)
print('25'*10)      # "*" * 10
print(25/10)
print(25//10)
print(25%10)
```

### 파이썬의 문자열은 문자의 시퀸스로서 불변(immutable)의 특성을 갖는다.
- 큰 타옴표("), 작은 따옴표(') 모두 사용 가능
- 여러 행으로 구성되는 문자열에는 ''' 또는 """ 을 사용  

```python
v1 = '둘리'
v2 = '또치'
v3 = "도우너"

print(v1)
print(v2)
print(v3)
```

```python
print(v1, end="")
print(v2)
print(v2, v3)
print(v2, v1, v3, sep="")
```

### 함수 : 임의의 기능을 지원하는 코드 블록, 이 함수의 기능이 필요할 때 호출이라는 방법으로 사용
### 내장함수 : 파이썬 언어에서 제공하는 함수
### print() : () 안에 작성된 데이터를 화면에 출력하는 기능을 제공하는 파이썬의 내장함수

```python
print(v1 + v2 + v3) #둘또도
print(v1, v2, v3) # 둘 또 도
print(v1, v2, v3, sep='') # 둘또도
print(v1, v2, v3, sep=',') # 둘, 또, 도
print(v1, v2, v3, sep='$$$') # 둘$$$또$$$도
```

```python
su1 = input('숫자 1 : ')
su2 = input('숫자 2 : ')
print('타입변환 전....')
print(type(su1))
print(type(su2))
print('타입변환 후....')
su1 = int(su1)
su2 = int(su2)
print(type(su1))
print(type(su2))
```

### <span style="color:red">파이썬의 대입 연산</span>

```python
v = 10
```

```python
a = b = c = 100
```

```python
print(v, a, b, c)
```

```python
x = 30
y = 50

print('x = ', x, ', y = ', y, sep="")
```

```python
print(f'x = {x}, y = {y}')  # f-string
```

```python
y, x = x, y
```

```python
print(f'x = {x}, y = {y}')
```

```python
v1, v2, v3, v4 = 11, 22, 33, 44
```

```python
print(v1, v2, v3, v4)
```

```python
v1, *v2 = 11, 22, 33, 44
```

```python
print(v1, v2)
```

```python
v1, *v2, v3 = 11, 22, 33, 44
```

```python
print(v1, v2, v3)
```

```python
v1, v2, v3 = 11, 22, 33, 44  # 오류 발생 - too many values to unpack (expected 3)
```

```python
print(v1, v2, v3)
```
