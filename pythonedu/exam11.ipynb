```python
def f1(n) :
    return n * 2

def f2() :
    print("f2 호출 - hello")

print(type(f1))
print(type(f2))
print(f1(10))
print(f2())
```

```python
def say1() :
    print("안녕?")

def say2() :
    print("Hello?")

import types
def callback(fn) :
    if type(fn) ==  types.FunctionType :
        fn()
    else :
        print("아규먼트로 함수를 주세요")

callback(say1)
callback(say2)
callback(100)
```

```python
def half(s):
    return s / 2

score = [ 45, 89, 72, 53, 94 ]
for s in map(half, score): 
    #score의 모든 요소를 half함수에 넣어서 결과를 반환하라
    print(s, end = ', ')
```

# <span style='color:blue; font-weight:bold'>예외처리</span>

### 문장이나 표현식이 구문적으로 올바르더라도 <mark>런타임에 발생하는 오류는 예외</mark>라고 함 
### <span style="color:red; font-weight:bold">예외란 실행 중에 감지된 실행 오류</span>

#### 예외 처리(exception handling)란, 프로그래머가 예기치못한 예외의 발생에 <mark>미리 대처하는 코드를 작성하는 것</mark>으로, 실행중인 프로그램의 비정상적인 종료를 막고, 상태를 정상상태로 유지하는 것이 목적이다.

```python
str_v = "89점"
score = int(str_v)
print(score)
print("작업완료")
```

```python
str_v = "89"
try :
	score = int(str_v)
	print(score)
except :
	print("예외가 발생했습니다.")
print("작업완료")
```

```python
while True:
    score = input("점수를 입력하세요 : ")
    try :
        score = int(score)
        print("입력한 점수 :", score)
        break
    except :
        print("점수 형식이 잘못되었습니다. 다시 합시다!")
    finally :
        print("난 항상 수행하는 코드입니다.")
print("작업완료")
```

```python
input_data = input("입력 : ")
try:
	score = int(input_data)
	print(score)
	elem = input_data[3]
	print(elem)
except ValueError :
	print("~~점수의 형식이 잘못되었습니다.~~")
except IndexError as e :
	print("인덱스 범위를 벗어났습니다.!!!:", e)
else :
    print("예외가 발생하지 않고 잘 수행했어요(^^)")
finally :
    print("예외 발생 여부 관계없이 항상 수행해요.")
print("작업완료")
```

```python
import random
num = random.randint(0,2) # 0 1 2
try :
	print("네트워크 접속")
	print(10 / num)
	print("네트워크 통신 수행")
except ZeroDivisionError as z :
	print(z)
finally :
	print("접속 해제")
print("작업 완료")
```

```python
import random, sys
num = random.randint(0,2) # 0 1 2
try :
	print("네트워크 접속")
	print(10 / num)
	print("네트워크 통신 수행")
except :
    exec_type, exec_value, exec_traceback = sys.exc_info()
    print(f"예외 타입: {exec_type}")
    print(f"예외 메시지: {exec_value}")
    print(f"예외 추적 정보: {exec_traceback}")
    print(f"{exec_traceback.tb_lineno}행에서 오류 발생")
finally :
	print("접속 해제")
print("작업 완료")
```

```python
def even_odd_num_checker(number):
	if not isinstance(number, int):
        #number가 int인지 검사 True또는 else
		raise TypeError("'number' is not an integer.")
	if number <= 0:
		raise ValueError("'number' must be positive.")
	if number % 2 == 0:
		print("This number is EVEN")
	else:
		print("This number is ODD")
```

```python
class CustomError(Exception):
    def __init__(self, message, error_code):
        self.message = message
        self.error_code = error_code
```

```python
try:
    raise CustomError("뭔가 문제가 있음~~", 500)
except CustomError as e:
    print(f"에러 메시지 : {e.message}")
    print(f"에러 코드 : {e.error_code}")
```

### <span style='color:blue; font-weight:bold'>Context Manager(컨텍스트 매니저) 란?</span>
#### 컨텍스트 매니저는 명령문 내에서 실행되는 런타임 컨텍스트를 정의하는는 객체
#### 코드 블록을 진입하고 나올 때 자동으로 설정 및 정리 작업을 수행하는데 사용
#### 자원 관리(파일, 네트워크 연결 등)에 유용
#### 파이썬은 쉬운 리소스 관리를 지원하기 위해 컨텍스트 매니저를 제공하며 <span style='color:red; font-weight:bold'>with 문</span>을 통해서 구현

## <span style="color:blue; font-weight:bold">with 문</span>

### with open(파일 경로, 모드) as 변수 :

```python
try:
    with open("output/live.txt", "r", encoding="UTF-8") as f :
        text = f.read()
        print(text)
except FileNotFoundError:
    print("파일이 없습니다.")
```
