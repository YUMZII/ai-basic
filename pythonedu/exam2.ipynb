```python
import random
num = random.randint(1, 20)
if num % 2 == 0 :
	print(num, "은 짝수", sep="")
else :
	print(num, "은 홀수", sep="")
```

```python
num = random.randint(1, 20)
if num % 2 == 0 :
	print(num)
	print("은 짝수")
else :
	print(num)
	print("은 홀수")
```

```python
num = random.randint(1, 20)
if num % 2 == 0 :
	print(num, end="")
	print("은 짝수")
else :
	print(num, end="")
	print("은 홀수")
```

```python
menu = int(input('메뉴를 입력하세요(1~3)> '))
 
match menu:  
    case 1:
        print('커피')
    case 2:
        print('에이드')
    case 3:
        print('우유')
    case _:  # _ 는 와일드카드라고 부르며 앞에 조건외의 모든을 의미, 제일 아래에 있어야 함
        print('1 ~ 3 까지만 입력 가능!')
   
```

```python
grade = input('학년을 입력하세요> ')
 
match grade:
    case '1' | '2' | '3' :
        print(grade + '학년은 저학년입니다.')
    case '4' | '5' | '6' :
        print(f'{grade}학년은 고학년입니다.')    
    case _ :  # _ 는 와일드카드라고 부르며 앞에 조건외의 모든을 의미
        print('1 ~ 6 까지만 입력 가능!')
```
