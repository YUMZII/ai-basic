```python
# 1부터 30까지의 정수숫자들중에서 3의 배수는 "3의 배수입니다."
# 아니면 "3의 배수가 아닙니다."를 출력하자.
for v in range(1,31):
	if v % 3 == 0 :
		print(v, ": 3의 배수입니다.")
	else :
		print(v, ": 3의 배수가 아닙니다.")
print(v) # 파이썬은 블럭 스코프를 지원하지 않음, 그래서 마지막 v 값이 출력됨
```

```python
for dan in range(1, 10):
	for num in range(1, 10):
		result = dan * num
		print(dan, "*", num, "=", result, end="\t")
	print()
```

```python
for dan in range(1, 10):
	for num in range(1, 10):
		result = dan * num
		if result > 30 :
			break   # 반복문을 끝내라
		print(dan, "*", num, "=", result, end="\t")
	print()
```

```python
endFlag = False
for dan in range(1, 10):
	for num in range(1, 10):
		result = dan * num
		if result > 30 :
			endFlag = True
			break
		print(dan, "*", num, "=", result, end="\t")
	if endFlag :
		break
	print()
```

```python
munja = 65  # A 의 코드값(10진수)
for y in range(1, 10) :
	for x in range(y) :
		print(chr(munja), end = '')
		munja += 1
	print()
```

```python
munja = 0x41  # A 의 코드값(16진수)
for y in range(1, 10) :
	for x in range(y) :
		print(chr(munja), end = '')
		munja += 1
	print()
```

```python
munja = 0b01000001 # A 의 코드값(2진수) - 10진수로는 65 16진수로는 0x41
for y in range(1, 10) :
	for x in range(y) :
		print(chr(munja), end = '')
		munja += 1
	print()
```

```python
munja = 44032 # 가 0xAC00
for y in range(1, 10) :
	for x in range(y) :
		print(chr(munja), end = '')
		munja += 1
	print()
```
