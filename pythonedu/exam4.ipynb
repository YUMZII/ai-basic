```python
import random
dice_num = random.randint(1, 6)
print("추출된 주사위 값 :", dice_num)
while dice_num >= 4 :
	print("ㅋㅋ")
	dice_num = random.randint(1, 6)
	print("추출된 주사위 값 :", dice_num) #여기서 추출된 dice_num을 위에 while문에서 그대로 사용
print("종료!!")
```

```python
while True :
	dice_num = random.randint(1, 6)
	print("추출된 주사위 값 :", dice_num)
	if dice_num >= 4 :
		print("ㅋㅋ")
	else :
		break

print("종료!!")
```

```python
print("3 + 4 = ?")
while True:
    a = input("정답을 입력하시오 : ")
    if a == "" :
    	print("입력된 값이 없어요")
    	continue
    if int(a) == 7 :
    	break        
    print("정답이 아니네요...")
print("참 잘했어요")
```

```python
dice_num = 1  # 초기화
count = 0
while dice_num != 3:    # 3이 아닐 때 계속 반복
	dice_num = random.randint(1, 6)    # randint를 사용하여 1과 6 사이의 난수를 생성
	print(dice_num)
	count += 1
print(count, "회 반복", sep="")
```

```python
count = 0
while True:
	count += 1
	num = random.randint(1, 20)    
	if num % 5 == 0 :
		print("끝낼께용!!")
		break
	elif num % 2 == 0 :
		print("다시 할께용!!")
		continue
	else :
		for _ in range(num) :
			print("#", end="")
		print()
	print("지금까지", count, "회 반복")
print("====== 수행 종료 =====")
```

```python
money = 1000
print("<<<현재 용돈은 ", money, "원입니다.~~ 소비시작>>>")
while True :
    pay = random.randint(100, 300)        
    if money > pay:
        money -= pay
        print(pay, "를 소비합니다.(ㅎㅎ)")
    else:
        print("현재 용돈은 ", money, "원이므로", pay, "원 만큼의 소비는 불가합니다.(ㅜㅜ)")
    if money < 100:
        print("현재 용돈은 ", money, "원이라서 더이상 소비는 불가합니다.(ㅜㅜ)")
        break
    print("현재 용돈은 ", money, "원입니다.")
```

### 파이썬 제어문의 종류
1. if문 (특정 조건의 만족 여부에 따라 실행)
2. for문 (특정 횟수만큼 반복 실행)
3. while문 (특정 조건을 만족하는 동안 반복 실행)

```python
print("*입력한 숫자만큼 데코문자를 출력하는 프로그램입니다.*")
deco = input("데코문자를 입력하세요 : ")
number = int(input("출력횟수를 입력하세요 : "))
for i in range(number) :
	print(deco, end="")
print("\n*수행 종료됩니다.*")
```

```python
print("*입력한 숫자만큼 데코문자를 출력하는 프로그램입니다. 3번 반복!!*")
for i in range(3) :
	deco = input("데코문자를 입력하세요 : ")
	number = int(input("출력횟수를 입력하세요 : "))
	for i in range(number) :
		print(deco, end="")
	print()
print("*수행 종료됩니다.*")
```

```python
print("*입력한 숫자만큼 데코문자를 출력하는 프로그램입니다. 원할 때까지 반복합니다.*")
while True :
	deco = input("데코문자를 입력하세요 : ")
	number = int(input("출력횟수를 입력하세요 : "))
	for i in range(number) :
		print(deco, end="")
	print()
	aws = input("계속 수행할까요(y/n)? ")
	if aws == 'n' :
		break
print("*수행 종료됩니다.*")
```

```python
print("*입력한 숫자만큼 데코문자를 출력하는 프로그램입니다. 원할 때까지 반복합니다.*")
while True :
	deco = input("데코문자를 입력하세요 : ")
	number = int(input("출력횟수를 입력하세요 : "))
	if number <= 0 :
		print("출력횟수는 0보다 큰 숫자를 입력하세요! ")
		continue
	for i in range(number) :
		print(deco, end="")
	print()
	aws = input("계속 수행할까요(y/n)? ")
	if aws == 'n' :
		break


        
print("*수행 종료됩니다.*")
```
