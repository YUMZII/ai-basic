```python
sum = 0
for num in range(5):
	sum += num
print("~ 4 =", sum)
```

```python
sum = 0
for num in range(11):
	sum += num
print("~ 10 =", sum)
```

```python
def calcsum(n):
	sum = 0
	for num in range(n + 1):
		sum += num
	return sum

print("~ 4 =", calcsum(4))
print("~ 10 =", calcsum(10))
```

```python
def calcrange(begin, end):
	sum = 0
	for num in range(begin, end + 1):
		sum += num
	return sum

print("3 ~ 7 =", calcrange(3, 7))
```

```python
def printsum(n):
	sum = 0
	for num in range(n + 1):
		sum += num
	print("~", n, "=", sum)

printsum(4)
printsum(10)
```

```python
def isType(p) :
	if type(p) == int :
		result = "정수를 전달했네요"
	elif type(p) == float :
		result = "실수를 전달했네요"
	elif type(p) == str :
		result = "문자열을 전달했네요"
	elif type(p) == bool :
		result = "논리값을 전달했네요"
	else :
		result = "몰라용^^"
	return result
```

```python
print(isType(100))
print(isType(3.14))
print(isType("100"))
print(isType("aaa"))
print(isType(True))
print(isType([])) #리스트
```

```python
def calcsum(n):
	"""n : 숫자 
       1 ~ n까지의 합계를 구해 리턴한다."""  # docstring
	sum = 0
	for i in range(n+1):
			sum += i
	return sum

print(calcsum(10))
```
