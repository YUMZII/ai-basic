## 리스트

```python
# 리스트 자료형(순서O, 중복O, 수정O, 삭제O)
a = []
b = list()
c = [1, 2, 3, 4]
d = [10, 100, 'Pen', 'Cap', 'Plate']
e = list('유니코')

print("리스트데이터의 타입")
print(type(d), d, sep="--->")
```

```python
print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
print("리스트데이터의 길이")
print(len(a))
print(len(d))
```

```python
print("리스트데이터의 인덱싱")
print('d - ', d[1])
print('d - ', d[0] + d[1] + d[1])
print('d - ', d[-1])
```

```python
# 슬라이싱
print('d - ', d[0:3])
print('d - ', d[2:])
```

```python
# 슬라이싱
print('d - ', d[:])
print('d - ', d[::-1])
```

```python
print("-- 리스트에 데이터 추가 --")
nums = [1, 2, 3, 4]
print(nums)
nums.append(5)
print(nums)
nums.insert(2, 99)
print(nums)
```

```python
print("-- 리스트에 범위를 지정하여 데이터 추가 --")
nums = [1, 2, 3, 4]
print(nums)
nums[2:2] = [90, 91, 92] # 세번째 위치에 요소 값들만 삽입되어 확장됨
print(nums)
```

```python
nums = [1, 2, 3, 4]
nums[2] = [90, 91, 92] # 세번째 위치에 리스트가 삽입됨
print(nums)
```

```python
print("-- 리스트 뒤에 또 다른 리스트 데이터 확장 --")
list1 = [1, 2, 3, 4, 5]
list2 = [10, 11]
list1.extend(list2)
print(list1)
```

```python
print("-- 리스트의 데이터를 삭제하는 다양한 방법 --")
score = [ 88, 95, 70, 100, 99, 80, 78, 50 ]
score.remove(100)
print(score)
del score[2]
print(score)
```

```python
score[1:4] = []
print(score)
```

```python
print("-- pop() 메서드로 리스트 데이터 지우고 꺼내기 --")
score = [ 88, 95, 70, 100, 99 ]
print(score.pop())
print(score.pop())
print(score.pop(1))
print(score)
```

```python
print("-- 리스트에서 데이터 위치(인덱스) 찾기와 갯수 카운팅 --")
score = [ 88, 95, 70, 100, 99, 80, 78, 50 ]
perfect = score.index(100)
print("만점 받은 학생은 " + str(perfect) + "번입니다.")
pernum = score.count(100)
print("만점자 수는 " + str(pernum) + "명입니다")
print("소팅전 :", score)
score.sort()
print("소팅후 :", score)
```

## 튜플

```python
# 튜플 자료형(순서O, 중복O, 수정X,삭제X)
# 선언
a = ()
b = (1,) # (1)
c = (1, 2, 3, 4)
d = (10, 100, 'Pen', 'Cap', 'Plate')
print('a - ', type(a), a)
print('b - ', type(b), b)
print('c - ', type(c), c)
print('d - ', type(d), d)
```

```python
# 인덱싱
print('d - ', d[1])
print('d - ', d[0] + d[1] + d[1])
print('d - ', d[-1])
```

```python
# 슬라이싱
print('d - ', d[0:3])
print('d - ', d[2:])
```

```python
# 튜플 연산
print('c + d - ', c + d)
print('c * 3 - ', c * 3)
print("'hi' + c[0] - ", 'hi' + str(c[0]))
```

```python
# 튜플 함수
a = (5, 2, 4, 3, 1, 4)

print('a - ', a)
print('a - ', a.index(5))
print('a - ', a.count(4))
```

```python
tuple_test = 10, 20, 30, 40
print("# 괄호가 없는 튜플의 값과 자료형 출력")
print("tuple_test:", tuple_test)
print("type(tuple_test):", type(tuple_test))
print()
```

```python
# 괄호가 없는 튜플 활용
a, b, c = 10, 20, 30
print("# 괄호가 없는 튜플을 활용한 할당")
print("a:", a)
print("b:", b)
print("c:", c)
```

```python
import time
def gettime():
    now = time.localtime()
    print(now, type(now))
    return now.tm_hour, now.tm_min
```

```python
result = gettime()
print(type(result))
print("지금은 %d시 %d분입니다" % (result[0], result[1]))
```
