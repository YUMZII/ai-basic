## 세트

```python
# 세트 자료형(순서X, 중복X, 수정O, 삭제O)
a = set()
b = set([1, 2, 3, 3, 4]) # iterable 객체가 와야 함
c = {1, 4, 5, 6, 1, 4}
d = set([1, 2, 'Pen', 'Cap', 'Plate'])
e = set((10,))
f = {100,}
print('a - ', type(a), a)
print('b - ', type(b), b)
print('c - ', type(c), c)
print('d - ', type(d), d)
print('e - ', type(e), e)
print('f - ', type(f), f)
```

```python
# for 문으로 데이터 추출
for data in d :
    print(data)
```

```python
# 튜플 로 변환
t = tuple(b)
print('tuple - ', type(t), t)
print('tuple - ', t[0], t[1:3])
```

```python
# 리스트로 변환
ldata = list(c)
print('list - ', type(ldata), ldata)
print('list - ', ldata[0], ldata[1:3])
```

```python
print("----set 의 합집 연산----")
s1 = set([1, 2, 3, 4, 5, 6])
s2 = set([4, 5, 6, 7, 8, 9])
print(s1, s2)
print('intersect - ', s1 & s2) #4,5,6
print('union - ', s1 | s2) #{1, 2, 3, 4, 5, 6, 7, 8, 9}
print('difference - ', s1 - s2) # 1,2,3
print('exclusive - ', s1 ^ s2) #{1, 2, 3, 7, 8, 9}
```

```python
print("----추가 & 제거 & 갯수 채크----")
s1 = set()
print(s1)
s1.add(10);s1.add(20);s1.add(30);s1.add(40);s1.add(50)
print('s1(10,20,30,40,50추가) - ', s1)
```

```python
s1.add(10)
print('s1(10추가실패) - ', s1)
```

```python
s1.update([40,50,60,70])
print('s1(40,50,60,70변경) - ', s1)
```

```python
s1.remove(30)
print('s1(30삭제) - ', s1)
print('length - ', len(s1))
```

```python
mammal = { '코끼리', '고릴라', '사자', '고래', '사람', '원숭이', '개' }
primate = { '사람', '원숭이', '고릴라' }

if '사자' in mammal:
    print("사자는 포유류이다")
else:
    print("사자는 포유류가 아니다.")
```

```python
print(primate & mammal)
print(primate | mammal)
print(primate ^ mammal)
```

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8, 9}

print(a & b)
print(a | b)
print(a ^ b)
```

## 딕셔너리

```python
dic_v = { 'boy':'소년', 'school':'학교', 'book':'책' , 'student':'나'}
print(dic_v)

print(dic_v['boy'])
print(dic_v['book'])
```

```python
print(dic_v['student']) # 오류 발생
```

```python
print(dic_v.get('student2'))
print(dic_v.get('student2', '사전에 없는 단어'))
```

```python
if 'student' in dic_v:
    print("사전에 있는 단어입니다.")
else:
    print("이 단어는 사전에 없습니다.")
```

```python
dic_v['boy'] = '남자애'
dic_v['girl'] = '소녀'
```

```python
del dic_v['book']
print(dic_v)
```

```python
dic_v = { 'boy':'소년', 'school':'학교', 'book':'책' }
print(dic_v.keys())
print(dic_v.values())
print(dic_v.items())
```

```python
for data in dic_v:
    print(data)
```

```python
for key in dic_v.keys():
    print(key)
```

```python
for value in dic_v.values():
    print(value)
```

```python
for item in dic_v.items():
    print(item)
```

```python
for key,value in dic_v.items():
    print(key, value, sep="-")
```
