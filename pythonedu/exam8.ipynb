# <span style="color:blue;font-weight:bold">8. 시퀀스 자료형(sequence type)</span>

### 파이썬에서는 리스트, 튜플, range, 문자열처럼 데이터 값들이 연속적으로 
### 이어진 데이터를 다루는 자료형을 <span style="color:red">시퀀스 자료형(sequence type)</span>라고 한다.

### <span style="color:blue; font-weight:bold">문자열</span>

```python
s1 = "abcdefgh1abcaa234가나다"
print(s1)
```

```python
print("---in과 not in 연산자---")
print('a' in s1)
print('a' not in s1)
print('가나' in s1)
print('가다' not in s1)
```

```python
print("---시퀀스연산---")
s1 = "abcdefgh1abcaa234가나다"
s2 = "파이썬"
print(s1+s2) #TypeError: can't multiply sequence by non-int of type 'str'
print(s1)
print(s2)
print(s2 * 3)
print(s2.__mul__(3))
```

```python
print("---인텍싱과 슬라이싱---")
s1 = "abcdefgh1abcaa234가나다"
s2 = "파이썬"
print(s2[0]); print(s2[1]); print(s2[2]); print(s2[-1])
#파
#이
#썬
#썬
print(s1[:]) #abcdefgh1abcaa234가나다
print(s1[::1]) #abcdefgh1abcaa234가나다
print(s1[::2]) #aceg1ba24나
print(s1[::-1]) #s1거꾸로
print(s1[0:4:1]) #abcd
print(s1[0:4]) #abcd
```

```python
s1, s2, s3 = "가나다"
print(s1)
print(s2)
print(s3)
```

```python
print(s1)
```

```python
print("---함수들---")
print(len(s1))
print(max(s1))
print(min(s1))
print(s1.count('a'))
print(s1.count('abc'))
```

```python
print("---for와 함께 사용하는 시퀀스---")
for data in s2 :
    print("#", data, "#")
```

```python
print("---문자열은 변경불가---")
s2[0] = "가" # 오류 발생
```

### <span style="color:blue; font-weight:bold">리스트</span>

```python
list1 = [10, 20, 30, "가나다", "abc", "A", True, 10, 20, 10]
print(list1)
```

```python
print("---in과 not in 연산자---")
print('30' in list1)
print(30 not in list1)
print(True in list1)
print(False not in list1)
```

```python
print("---시퀀스연산---")
list2 = ["파","이","썬"]
print(list1+list2)
print(list1)
print(list2)
print(list2 * 3)
```

```python
print("---인텍싱과 슬라이싱---")
print(list2[0]); print(list2[1]); print(list2[2]); print(list2[-1])
print(list1[:])
print(list1[::1])
print(list1[::2])
print(list1[::-1])
print(list1[0:4:1])
print(list1[0:4])
```

```python
e1, e2, e3 = [100, 'abc', True]
print(e1)
print(e2)
print(e3)
```

```python
print("---함수들---")
print(len(list1))
#print(max(list1)); print(min(list1))
print(max([10, 20, 30])); print(min([10, 20, 30]))
print(max(['a', 'b', 'c'])); print(min(['a', 'b', 'c']))
print(max([True, False])); print(min([True, False]))
print(list1.count(10))
print(list1.count(20))
```

```python
print("---for와 함께 사용하는 시퀀스---")
for data in list2 :
    print("#", data, "#")
```

```python
print("---리스트는 변경가능---")
list2[0] = "가"
print(list2)
```

## <span style='color:blue; font-weight:bold'>모듈(module)</span>

- 하나의 파이썬 소스(.py)
- 함수, 클래스, 변수 등을 포함
- 다른 곳(xxx.py, xxx.ipynb)에서 import 구문으로 사용 가능

## 패키지/라이브러리

모듈이 함수의 집합이라면 패키지(package)는 모듈을 모아놓은 <span style="color:red">폴더</span>라고 할 수 있다. 패키지는 종종 라이브러리(library)라고도 부른다. 엄말하게 말하면 라이브러리는 패키지의 집합으로 패키지 보다 포괄적인 개념이지만, 혼용해서 사용하기도 한다. 

예를 들어, 넘파이 라이브러리라고도 부르고, 넘파이 패키지라고도 부른다.

```python
print("파이썬 인터프리터가 직접 실행시키는 파이썬 소스에 자동으로 부여되는 모듈명 :", __name__)
```
