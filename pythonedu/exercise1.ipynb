```python
num5 = random.randint(0,100)
match num5//10:
    case 10 | 9:
        print(f'{num5} 점은 A등급입니다.')
    case 8:
        print(f'{num5} 점은 B등급입니다.')
    case 7:
        print(f'{num5} 점은 C등급입니다.')
    case 6:
        print(f'{num5} 점은 D등급입니다.')
    case _:
        print(f'{num5} 점은 F등급입니다.')
```
