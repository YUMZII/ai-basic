```python
#5
import random
start = random.randint(1,10)
end = random.randint(30,40)
sum=0
for i in range(start, end, 1):
    if i%2==0:
        sum += i
print(f'{start} 부터 {end} 까지의 합: {sum}')
```

```python
#6
even_num = 0
odd_num = 0
for i in range(1, 101):
    if i%2==0:
        even_num += i
    else:
        odd_num += i
print('1부터 100까지의 숫자들 중에서')
print(f'짝수의 합은 {even_num}')
print(f'홀수의 합은 {odd_num}')
```

```python
#8
while True:
    pairNum1 = random.randint(1,6)
    pairNum2 = random.randint(1,6)

    if pairNum1 > pairNum2:
        print("pairNum1 이 pairNum2 보다 크다.")
    elif pairNum1 < pairNum2:
        print("pairNum1 이 pairNum2 보다 작다.")
    else:
        print("게임 끝")
        break
```

```python
#9
import random
count = 0 
while True:
    ran2 = random.randint(0,30)
    count += 1
    if ran2 == 0 or ran2 >= 27:
        break
    print(f'{chr(ran2 + 64)}({ran2})')
print(f"수행횟수는 {count}번입니다.")
```

```python
#10
while True:
    ans = int(input("1~12사이의 값을 입력하세요"))
    if ans in (12,1,2):
        print(f'{ans}월은 겨울')
    elif 3 <= ans <= 5:
        print(f'{ans}월은 봄')
    elif 6 <= ans <= 8:
        print(f'{ans}월은 여름')
    elif 9 <= ans <= 11:
        print(f'{ans}월은 가을')
    else:
        print("1~12 사이의 값을 입력하세요!")
        break
```
