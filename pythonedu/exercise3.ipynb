```python
#3
def expr(num1,num2,an):
    if an == '+':
        answer = num1 + num2
    elif an == '-':
        answer = num1 - num2
    elif an == '*':
        answer = num1 * num2
    elif an == '/':
        answer = num1 / num2
    else:
        print("수행불가")
        return
    print(f"연산결과: {answer}")
    return answer
```

```python
print(expr(1,2,'*'))
```

```python
#4
def print_triangle1(num3):
    if num3 > 11:
        return
    else:        
        for i in range(1, num3 + 1):
            for j in range(i):
                print("*", end="")
            print()
```

```python
print_triangle1(5)
```

```python
print_triangle1(14)
```

```python
#5
def print_gugudan(n):
    if type(n) != type(1) :
        return
    else:
        for i in range(1,n):
            for j in range(1,10):
                result = i * j
                print(i, "x", j, "=", result, end=" ")
                print()
    return
```

```python
print_gugudan(2)
```

```python
print_gugudan('SK')
```

```python
#6
def differ_two_value(num1, num2):
    max1 = max(num1, num2)
    min1 = min(num1, num2)
    result = max1 - min1
    return result
```

```python
import random
for _ in range(5):
    ran = random.randint(1,31)
    ran1 = random.randint(1,31)
    differ_two_value(ran, ran1)
    print(f'{ran}와 {ran1}의 차는 {differ_two_value(ran, ran1)}')
```
