```python
#4
listnum = [[10, 20, 30, 40, 50], [5, 10, 15],[11, 22, 33, 44], [9, 8, 7, 6, 5, 4, 3, 2, 13
]]
for idx, row in enumerate(listnum, start=1):
    total = 0
    for num in row:
        total += num
    print(f'{idx}행의 합은{total}입니다')
```

```python
#5
listst = [['B', 'C', 'A', 'A'], ['C', 'C', 'B', 'B'],
          ['D', 'A', 'A', 'D']]

count1 = 0
count2 = 0
count3 = 0
count4 = 0
for row in listst:
    for ch in row:
        if ch == 'A':
            count1 += 1
        elif ch == 'B':
            count2 += 1
        elif ch == 'C':
            count3 += 1
        elif ch == 'D':
            count4 += 1
print(f'A는 {count1}개 입니다.')
print(f'B는 {count2}개 입니다.')
print(f'C는 {count3}개 입니다.')
print(f'D는 {count4}개 입니다.')
```

```python
#6
color_dic = {
    "red":"#FF0000",
    "blue":"#0000FF",
    "green":"#008000",
    "yellow":"#FFFF00",
    "orange":"#FFA500",
    "black":"#000000",
    "white":"#FFFFFF",
    "violet":"#EE82EE",
    "pink":"#FFC0CB",
    "lime":"#00FF00"
}

color = input("칼라명을 영문으로 입력하세요 : ")

if color in color_dic:
    print(f"{color} 칼라의 RGB 값은 {color_dic[color]} 입니다")
else:
    print(f"{color} 칼라의 RGB 값을 찾을 수 없습니다")
```

```python
#7
weather = {
    "목": (16, 27),
    "금": (16,28),
    "토": (17,30),
    "일": (19,30)
}

day = input("요일명을 한글로 입력하세요 : ")

if day in weather:
    low, high = weather[day]
    print(f"{day}요일의 최저온도는 {low} 이고 최고 온도는 {high}입니다")
else:
    print(f"{day}요일의 정보를 찾을 수 없습니다")
```

```python
#9
import random

set1 = set()
set2 = set()

while len(set1) < 10:
    set1.add(random.randint(1,20))

while len(set2) < 10:
    set2.add(random.randint(1,20))

print("집합1 :", set1)
print("집합2 :", set2)
print("두 집합에 모두 있는 데이터 :", set1 & set2)
print("집합1 또는 집합2 에 있는 데이터 :", set1 | set2)
print("집합1에는 있고 집합2에는 없는 데이터 :", set1 - set2)
print("집합2에는 있고 집합1에는 없는 데이터 :", set2 - set1)
print("집합1과 집합2가 각자 가지고 있는 데이터 :", set1 ^ set2)
```
